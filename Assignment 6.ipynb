## Q4

```python
X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])

y = np.array([40, 65, 30, 85])
```

```python
print(X.shape)
print(X.ndim)
```

```python
xTrans=X.T
print(xTrans)
```

```python
mul=xTrans@X
print(mul)
```

```python
inv=np.linalg.inv(mul)
print(inv)
```

```python
b=inv@xTrans@y
print(b)
```
