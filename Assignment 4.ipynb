## Q1

```python
rollNum="1024170222"
last=int(rollNum[-1])
secLast=int(rollNum[-2])
digits=[secLast,last]
```

```python
categories=["billing", "account", "general"]
fixedEntry = [ 
{"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", 
"keywords": "fee cost price charge", "category": "billing"}, 
{"question": "how to reset password", "answer": "Go to Settings > Reset Password.", 
"keywords": "password reset login", "category": "account"}, 
{"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", 
"keywords": "hours timing open time", "category": "general"}, 
{"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", 
"keywords": "pay payment upi fee", "category": "billing"}, 
]
```

```python
personalEntry=[]
for d in digits:
    cat=categories[d%3]
    if cat=="billing":
        entry = {
        "question": "where can i download my fee receipt",
        "answer": "You can download your fee receipt from the billing section.",
        "keywords": "receipt invoice billing",
        "category": cat
    }
    elif cat=="general":
        entry = {
        "question": "where is the DOAA office",
        "answer": "It's located in the admin block near B block.",
        "keywords": "DOAA office location",
        "category": cat
    }   
    else :
        entry = {
        "question": "How can I update my home address",
        "answer": "Go to settings -> update -> chage address",
        "keywords": "address update account",
        "category": cat
    }   
    personalEntry.append(entry)

totalEntry=fixedEntry+personalEntry
df = pd.DataFrame(totalEntry)
print(df)
```

## Q2

```python
def scoreQuery(query,df):
    queryWords = set(query.lower().split())
    res=[]
    for ind,row in df.iterrows():
        searchText=(row["question"].lower()+ " "+ row["keywords"].lower())
        searchWords = set(searchText.split())
        matchWords = queryWords.intersection(searchWords)
        score=len(matchWords)
        if score>0:
            res.append({
                "index": ind,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })
    res.sort(key=lambda x: x["score"],reverse=True)
    return res
```

```python
query="fee payment"
results= scoreQuery(query,df)
for res in results:
    print("\nQuestion:", res["question"])
    print("Answer:", res["answer"])
    print("Category:", res["category"])
    print("Confidence:", res["score"])
```

## Q6

```python
def scoreQueryTie(query,df):
    queWord=query.lower().split()
    res=[]
    for ind,row in df.iterrows():
        score=0
        text=(row["question"].lower() + " " +row["answer"].lower() + " " +row["keywords"].lower())
        for word in queWord:
            if word in text:
                score += 1
        if score > 0:
            res.append((row["question"], score))

    print("res =", res)
    if not res:
        print("No Match")
        return
    maxScore=max(score for _,score in res)
    best=[ 
        item for item in res
        if item[1]==maxScore
    ]
    print("Best Match")
    for ques,score in best:
        print("Question:", ques, " | Score:",score)

scoreQueryTie("DOAA",df)
print("\n")
scoreQueryTie("password",df)
print("\n")
scoreQueryTie("hostel",df)
```
