```python
def isPrime(n):
    for i in range(2,n//2+1):
        if n%i==0:
            return "NO"
    return "Yes"

print ("IsPrime(24)=", isPrime(24))
print ("IsPrime(56)=", isPrime(56))
print ("IsPrime(7)=", isPrime(7))
print ("IsPrime(63)=", isPrime(63))
```

```python
def addition(n):
    s=sum(range(n+1))
    return s
print("Addition upto 76:",addition(76))
```

```python
def addOdd(n):
    s=sum(range(1,n+1,2))
    return s
print("Addition of odd numbers upto 76:",addOdd(76))
```

```python
def addPrime(n):
    s=0
    for i in range(2,n//2):
        if n%i != 0:
            s+=i;
    return s
print("Addition ofprime numbers upto 76:",addPrime(76))
```

```python
print("Letters + Digits =", st.ascii_letters + st.digits)

password = r.sample(st.ascii_letters + st.digits, 6)
print("\nSelected List1 =", password)
password = "".join(password)
print("password1 =", password)

password = r.sample(st.ascii_letters + st.digits, 7)
print("\nSelected List2 =", password)
password = "".join(password)
print("password2 =", password)

symbols = "$&?@!*%+-_=<>/"
password = r.sample(symbols + st.ascii_letters + st.digits, 9)
print("\nSelected List3 =", password)
password = "".join(password)
print("password3 =", password)
```
