```python
arr= np.array([1, 2, 3, 6, 4, 5])
revArr=arr[::-1]
print("Original:",arr)
print("Reversed:",revArr)
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
maxFreq=np.bincount(x).argmax()
maxInd=np.where(x == maxFreq)[0]
print(maxFreq)
print(maxInd)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
maxFreq=np.bincount(y).argmax()
maxInd=np.where(y == maxFreq)[0]
print(maxFreq)
print(maxInd)
```
