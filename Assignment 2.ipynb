```python
import random
print(random.seed(102417022))

num=[random.randint(100, 900) for _ in range(100)]
print("Random List:",num)
```

```python
count=0;
for i in num:
    if i%2!=0:
        count=count+1
        print(i)
print("Total odd:",count)
```

```python
count=0;
for i in num:
    if i%2==0:
        count=count+1
        print(i)
print("Total even:",count)
```

```python
def isPrime(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i == 0:
            return False
    return True
    
listPrime=[]
count=0;
for i in num:
    if (isPrime(i)):
        count=count+1
        print(i)
        listPrime.append(i)
print("Total prime:",count)
print(listPrime)
```

```python
maxFreq=max(num,key=num.count)
freq=num.count(maxFreq)
print(f"Most frequent number: {maxFreq} with frequency: {freq}")
```

```python
roll=10241702
setA=set()
setB=set()
for i in str(roll):
    i=int(i)
    setA.add(7*i)
    setB.add(8*i)

print(setA)
print(setB)
```

```python
setA.union(setB)
```

```python
setA.intersection(setB)
```

```python
print(setA-setB)
print(setB-setA)
```

```python
print(setA.symmetric_difference(setB))
```

```python
print("A a subset of B?",setA.issubset(setB))
print("B a superset of A?",setB.issuperset(setA))
```

```python
myDict["age"]=myDict["age"]+1
print(myDict)
```

```python
dict1 = myDict.copy()
dict2 = myDict.copy()

removeBranch = dict1.pop("branch")
del dict2["branch"]

print("After pop():", dict1)
print("Removed value using pop():", removeBranch)
print("After del:", dict2)
```

pop() removes the key and returns its value, while del removes the key but does not return its value.

```python
friendDict = {
    "name": "Nishi",
    "roll_no": "10241715",
    "branch": "CSE",
    "age": 21,
    "city": "Barnala"
}

merged={**myDict, **friendDict}
print("\nMerged dictionary:", merged)
```
