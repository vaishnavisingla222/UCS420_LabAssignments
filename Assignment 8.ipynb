```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
text = [
    "Artificial intelligence is transforming healthcare and education.",
    "Machine learning helps computers learn from data.",
    "Healthcare technology uses artificial intelligence for diagnosis."
]

countVector = CountVectorizer()
bowMatrix = countVector.fit_transform(text)
print("Bag of Words Representation:")
print("Vocabulary:", countVector.get_feature_names_out())
print(bowMatrix.toarray())
```

```python
tfIdfVector = TfidfVectorizer()
tfIdfMatrix = tfIdfVector.fit_transform(text)
word = tfIdfVector.get_feature_names_out()
print(tfIdfMatrix.toarray())
```

```python
for i in range(len(text)):
    score = tfIdfMatrix[i].toarray().flatten()
    top = score.argsort()[::-1][:3]
    print(f"\nText {i + 1}: {text[i]}")
    for index in top:
        print(f"{word[index]}: {score[index]:.4f}")
```

## Q4

```python
from sklearn.metrics.pairwise import cosine_similarity

text1 = """
Artificial Intelligence enables machines to learn from data.
AI systems can solve problems and make intelligent decisions.
Machine learning is an important part of artificial intelligence.
AI is widely used in healthcare, education, and robotics.
"""

text2 = """
Blockchain is a decentralized digital ledger technology.
It stores transactions securely across multiple computers.
Blockchain provides transparency and prevents data tampering.
It is widely used in cryptocurrency and digital finance.
"""
```

```python
def preprocess(text):
    text = text.lower()
    tokens = re.findall(r'\b[a-z]+\b', text)
    return tokens

token1=preprocess(text1)
token2=preprocess(text2)
print("Tokens of Text 1:")
print(token1)
print("\nTokens of Text 2:")
print(token2)
```

```python
set1 = set(token1)
set2 = set(token2)
jaccard = len(set1.intersection(set2)) / len(set1.union(set2))
print(round(jaccard, 4))
```

```python
text=[text1, text2]
vector = TfidfVectorizer()
tfIdfMatrix = vector.fit_transform(text)
cos = cosine_similarity(tfIdfMatrix[0:1], tfIdfMatrix[1:2])[0][0]
print(round(cos, 4))
```

```python
print("Interpretation:")
if jaccard > cos:
    print("Jaccard Similarity is higher")
elif cos > jaccard:
    print("Cosine Similarity is higher")
else:
    print("Both similarity are equal.")
```

## Q5

```python
%pip install textblob wordcloud matplotlib
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "This smartphone has an excellent camera and amazing battery life. I love it.",
    "The headphones have poor sound quality and stopped working after two days.",
    "The laptop is average. It works fine but has no special features."
]
```

```python
posReview = []

print("Sentimental Analysis")
for i, review in enumerate(reviews, start=1):
    blob = TextBlob(review)
    polarity = blob.sentiment.polarity
    sub = blob.sentiment.subjectivity
    if polarity > 0:
        sentiment = "Positive"
        posReview.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(f"Review {i}: {review}")
    print(f"Polarity: {polarity:.2f}")
    print(f"Subjectivity: {sub:.2f}")
    print(f"Sentiment: {sentiment}\n")
```

```python
if posReview:
    posText = " ".join(posReview)
    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white"
    ).generate(posText)
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews available to generate a word cloud.")
```

```python
%pip install tensorflow
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

text = """
Artificial intelligence is changing the world through modern technology.
Machine learning helps computers learn from data and improve their performance.
Deep learning uses neural networks to solve complex problems.
Artificial intelligence is used in healthcare, education, finance, and robotics.
Technology makes our daily lives easier and more efficient.
Data plays an important role in building intelligent systems.
Researchers develop new algorithms to improve accuracy and speed.
Computers can recognize images, understand language, and make predictions.
Machine learning models require quality data for better results.
Artificial intelligence continues to create new opportunities for innovation.
Students can learn programming to build intelligent applications.
The future of technology depends on creativity, research, and continuous learning.
"""
```

```python
token= Tokenizer()
token.fit_on_texts([text])
total= len(token.word_index) + 1
print("Total unique words:", total)
print("Word Index:", token.word_index)

inpSeq = []

for sen in text.lower().split("."):
    sen = sen.strip()

    if sen:
        tokenList = token.texts_to_sequences([sen])[0]

        for i in range(2, len(tokenList) + 1):
            ngramSeq = tokenList[:i]
            inpSeq.append(ngramSeq)

# Pad sequences
maxSeqLen = max(len(seq) for seq in inpSeq)

inpSeq = pad_sequences(
    inpSeq,
    maxlen=maxSeqLen,
    padding="pre"
)

# Separate input and output
X = inpSeq[:, :-1]
y = inpSeq[:, -1]

# One-hot encode target words
y = np.eye(total)[y]

print("\nInput shape:", X.shape)
print("Output shape:", y.shape)
```

```python
model = Sequential([
    Embedding(total, 32, input_length=maxSeqLen - 1),
    LSTM(64),
    Dense(total, activation="softmax")
])
model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)
model.summary()
# Train the model
model.fit(X, y, epochs=150, verbose=1)
```

```python
def generate_text(seedText, nextWords=8):
    for _ in range(nextWords):
        tokenList = token.texts_to_sequences([seedText])[0]

        tokenList = pad_sequences(
            [tokenList],
            maxlen=maxSeqLen - 1,
            padding="pre",
            truncating="pre"
        )

        predictProbs = model.predict(tokenList, verbose=0)[0]
        predictInd = np.argmax(predictProbs)

        outWord = token.index_word.get(predictInd, "")

        if not outWord:
            break

        seedText += " " + outWord

    return seedText

# Generate 3 new lines
print("Generated Text:")
seed="artificial intelligence"

for i in range(3):
    print(f"Line {i + 1}: {generate_text(seed, nextWords=8)}")
```
